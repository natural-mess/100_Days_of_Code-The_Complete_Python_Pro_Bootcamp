### Data Type Conversions

**Challenge**: Convert the `USD_Production_Budget`, `USD_Worldwide_Gross`, and `USD_Domestic_Gross` columns to a numeric format by removing `$` signs and `,`. 
<br>
<br>
Note that *domestic* in this context refers to the United States.

````python
data.head()
````

````python
chars_to_remove = [',', '$']
for char in chars_to_remove:
    data.USD_Production_Budget = data.USD_Production_Budget.astype(str).str.replace(char, "", regex=False)
    data.USD_Worldwide_Gross = data.USD_Worldwide_Gross.astype(str).str.replace(char, "", regex=False)
    data.USD_Domestic_Gross = data.USD_Domestic_Gross.astype(str).str.replace(char, "", regex=False)

data.USD_Production_Budget = pd.to_numeric(data.USD_Production_Budget)
data.USD_Production_Budget
````

````python
data.USD_Worldwide_Gross = pd.to_numeric(data.USD_Worldwide_Gross)
data.USD_Worldwide_Gross
````

````python
data.USD_Domestic_Gross = pd.to_numeric(data.USD_Domestic_Gross)
data.USD_Domestic_Gross
````

**Challenge**: Convert the `Release_Date` column to a Pandas Datetime type. 

````python
data.Release_Date = pd.to_datetime(data["Release_Date"])
data.info()
````
